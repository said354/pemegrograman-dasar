```python

# Soal 1: Belanja Buku

# Input data
jumlah_buku = int(input("Masukkan jumlah buku: "))
harga_buku = int(input("Masukkan harga per buku: Rp "))
diskon_persen = 10
uang_budi = int(input("Masukkan uang Budi: Rp "))
jumlah_orang = 4

# Menghitung total belanja
total_belanja = jumlah_buku * harga_buku

# Menghitung diskon
diskon = total_belanja * diskon_persen / 100

# Menghitung total pembayaran
total_bayar = total_belanja - diskon

# Menghitung uang kembalian
kembalian = uang_budi - total_bayar

# Iuran per orang
iuran = int(total_bayar) // jumlah_orang

# Mengecek kecukupan uang
uang_cukup = uang_budi >= total_bayar

# Menampilkan hasil
print("\n=== HASIL PERHITUNGAN ===")
print("Total belanja  : Rp", total_belanja)
print("Diskon         : Rp", int(diskon))
print("Total bayar    : Rp", int(total_bayar))
print("Uang kembalian : Rp", int(kembalian))
print("Iuran per orang: Rp", iuran)
print("Uang cukup?    :", uang_cukup)
```

```python

# Soal 2: PLTS Atap Rumah

# Input data
jumlah_panel = int(input("Masukkan jumlah panel surya: "))
daya_panel = float(input("Masukkan daya per panel (Wp): "))
jam_matahari = float(input("Masukkan jam penyinaran per hari: "))
efisiensi = float(input("Masukkan efisiensi sistem (%): ")) / 100
kebutuhan_harian = float(input("Masukkan kebutuhan listrik (kWh/hari): "))
tarif_listrik = float(input("Masukkan tarif listrik per kWh (Rp): "))

# Menghitung total daya panel (kWp)
total_daya = jumlah_panel * daya_panel / 1000

# Menghitung energi yang dihasilkan per hari (kWh)
energi_harian = total_daya * jam_matahari * efisiensi

# Menghitung energi per bulan (30 hari)
energi_bulanan = energi_harian * 30

# Menghitung nilai rupiah listrik yang dihasilkan
nilai_listrik = energi_bulanan * tarif_listrik

# Menghitung kekurangan energi per hari
kekurangan = max(0, kebutuhan_harian - energi_harian)

# Mengecek apakah PLTS mencukupi kebutuhan
plts_cukup = energi_harian >= kebutuhan_harian

# Menampilkan hasil
print("\n=== HASIL PERHITUNGAN PLTS ===")
print(f"Energi per hari       : {energi_harian:.2f} kWh")
print(f"Energi per bulan      : {energi_bulanan:.2f} kWh")
print(f"Nilai listrik per bulan: Rp {nilai_listrik:,.2f}")
print(f"Kekurangan energi     : {kekurangan:.2f} kWh/hari")
print("PLTS mencukupi?       :", plts_cukup)
```

```python

# Soal: Generator Diesel

# Input data
daya_generator = float(input("Masukkan daya generator (kW): "))
jam_operasi = float(input("Masukkan jam operasi per hari: "))
konsumsi_solar = float(input("Masukkan konsumsi solar (liter/kWh): "))
harga_solar = float(input("Masukkan harga solar per liter (Rp): "))
kapasitas_tangki = float(input("Masukkan kapasitas tangki (liter): "))
arus = float(input("Masukkan arus kabel (A): "))
hambatan = float(input("Masukkan hambatan kabel (Ohm): "))

# 1. Energi yang dibangkitkan per hari
energi_harian = daya_generator * jam_operasi

# 2. Kebutuhan solar dan biaya per hari
solar_harian = energi_harian * konsumsi_solar
biaya_harian = solar_harian * harga_solar

# 3. Jumlah hari penuh tangki cukup dan sisa solar
hari_penuh = int(kapasitas_tangki // solar_harian)
sisa_solar = kapasitas_tangki - (hari_penuh * solar_harian)

# 4. Rugi daya pada kabel (P = I^2 x R)
rugi_daya = arus ** 2 * hambatan

# 5. Mengecek apakah rugi kabel tergolong tinggi
rugi_tinggi = (rugi_daya > 150) and (arus >= 20)

# Menampilkan hasil
print("\n=== HASIL PERHITUNGAN GENERATOR ===")
print(f"Energi per hari       : {energi_harian:.2f} kWh")
print(f"Kebutuhan solar       : {solar_harian:.2f} liter/hari")
print(f"Biaya solar per hari  : Rp {biaya_harian:,.2f}")
print(f"Tangki cukup untuk    : {hari_penuh} hari penuh")
print(f"Sisa solar            : {sisa_solar:.2f} liter")
print(f"Rugi daya kabel       : {rugi_daya:.2f} W")
print("Rugi kabel tinggi?    :", rugi_tinggi)
```
