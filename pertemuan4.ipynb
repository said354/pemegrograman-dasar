```python
# konversi waktu
total_detik = int(input("Masukkan total detik: "))

jam = total_detik // 3600
sisa_detik = total_detik % 3600
menit = sisa_detik // 60
detik = sisa_detik % 60

print(f"Hasil Konversi: {jam} Jam, {menit} Menit, {detik} Detik")
```

```python
tugas = float(input("Masukkan nilai tugas : "))
kuis = float(input("Masukkan nilai kuis : "))
praktikum = float(input("Masukkan nilai praktikum : "))
uts = float(input("Masukkan nilai UTS : "))
uas = float(input("Masukkan nilai UAS : "))

nilai_mahasiswa = (0.2 * tugas) + (0.1 * kuis) + (0.2 * praktikum) + (0.2 * uts) + (0.3 * uas)

standar_kelulusan = float(input("Masukkan standar kelulusan : "))

apakah_lulus = nilai_mahasiswa >= standar_kelulusan
perlu_remedial = nilai_mahasiswa < standar_kelulusan

print(f"Nilai Mahasiswa: {nilai_mahasiswa}")
print(f"Standar Kelulusan: {standar_kelulusan}")
print(f"Apakah Mahasiswa Lulus? {apakah_lulus}")
print(f"Perlu Remedial? {perlu_remedial}")
```

```python
# soal 1
belanja = int(input("Masukkan total belanja: "))
diskon = (0, 0.1 * belanja)[belanja > 500000]
total_harga = belanja - diskon
print("Total harga setelah diskon =", total_harga)


# soal 2
belanja = int(input("Masukkan total belanja: "))
diskon = (0, 0.35 * belanja)[belanja > 1000000]
total_harga = belanja - diskon
print("Total harga setelah diskon =", total_harga)
```

```python
# Latihan 1
tahun = int(input("Masukkan tahun: "))

if (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0):
    print(tahun, "adalah tahun kabisat")
else:
    print(tahun, "bukan tahun kabisat")
```

```python
# Latihan 2
# Program Hitung Luas dan Keliling Persegi Panjang

print("=== Kalkulator Persegi Panjang ===")

panjang = float(input("Masukkan panjang (cm): "))
lebar = float(input("Masukkan lebar (cm): "))

luas = panjang * lebar
keliling = 2 * (panjang + lebar)

print("Luas     =", luas, "cm2")
print("Keliling =", keliling, "cm")
```
