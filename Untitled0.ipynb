```python
#Jawaban 1
# Memasukkan dua bilangan
bilangan1 = 16
bilangan2 = 4

# Menampilkan hasil operasi
print("Penjumlahan:", bilangan1 + bilangan2)
print("Pengurangan:", bilangan1 - bilangan2)
print("Perkalian:", bilangan1 * bilangan2)
print("Pembagian:", bilangan1 / bilangan2)
print("Pembagian bulat:", bilangan1 // bilangan2)
print("Modulus:", bilangan1 % bilangan2)

#Jawaban 2
# Menentukan nilai ujian
nilai = 75

# Mengecek apakah nilai termasuk kategori lulus
lulus = nilai >= 60

print("Nilai ini termasuk kategori lulus (>= 60):", lulus)

#Jawaban 3
# Menentukan umur dan status SIM
umur = 19
punya_sim = True

# Mengecek apakah boleh mengemudi sendiri
boleh_mengemudi = (umur >= 17) and punya_sim

# Ekspresi dengan tanda kurung untuk latihan operator precedence
cek_syarat = (umur >= 17) and (punya_sim == True)

print("Boleh mengemudi sendiri:", boleh_mengemudi)
print("Hasil pengecekan syarat:", cek_syarat)

#Jawaban 4
# Menentukan energi input dan energi output
energi_input = 1000
energi_output = 850

# Menghitung efisiensi
efisiensi = (energi_output / energi_input) * 100

# Menampilkan hasil
print(f"Efisiensi termal unit pembangkit adalah {efisiensi:.2f}%")

#Jawaban 5
# Menentukan suhu boiler dan beban operasi
suhu_boiler = 560
beban_operasi = 90

# Mengecek kondisi maintenance
perlu_maintenance = (suhu_boiler > 550) or (beban_operasi > 95)

# Menampilkan hasil
print(f"Apakah unit pembangkit perlu maintenance? {perlu_maintenance}")
```
